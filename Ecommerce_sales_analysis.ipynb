**Data Extraction and Filtering**

```python
data["Customer"] # selecting the customers
```

```python
# Displays the details of orders that were sold at the max Quantity compared to all the items
data[(data["Quantity"] == max(data["Quantity"]))]
```

```python
# Displaying different products
data["Product"].unique()
```

```python
# Product with the highest unit price
data[data["Unit_Price"] == max(data["Unit_Price"])]["Product"].unique()
```

**Coffee Maker** has the highest unit price.

```python
# Orders between Jan 2026 and April 2026
data[(data["Order_Date"] >= "2026-01-01") & (data["Order_Date"] <= "2026-04-30")]
```

```python
# Displaying customers and their cities who bought sports items and Displayed in Ascending order of the customer
data[data["Category"] == "Sports"][["Customer","City"]].sort_values(by="Customer")
```

```python
# Displaying avg of Total_Sales of each Category
data.groupby("Category")["Total_Sales"].mean()
```

```python
# Displaying sum of Total_Sales of each Cities
data.groupby("City")["Total_Sales"].sum().sort_values()
```

**Observation** :
<br>
-  Books Category has the least average Total_Sales whereas Sports Category has the highest average Total_Sales.
-  Hyderabad City has the highest(sum) total sales.

```python
# Products and their max sales
data.groupby("Product")["Total_Sales"].max().sort_values(ascending = False)
```

**Observation** - Running Shoes has the highest Total_Sales

```python
# Payment method with Total_Sales
data.groupby("Payment_Method")["Total_Sales"].sum()
```

**Observation** - Most sales amount are done through Credit card

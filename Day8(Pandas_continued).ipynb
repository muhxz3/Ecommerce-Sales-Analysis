**Data selecting, extracting and slicing**

```python
import pandas as pd
```

```python
df = pd.DataFrame({
    "name": ['A','B','C','D'],
    "marks": [88,76,92,98],
    "city": ["Del","Pun","Koc","Mum"]
})

print(df)
```

```python
print(df["name"])
```

```python
print(df[["name","city"]])
```

**.loc and .iloc**<br>
.loc - label based<br>
.iloc - index position based

```python
# df.loc[rowlabel, columnlabel]
df.loc[1]
```

```python
df.loc[1,"name"]
```

```python
df.loc[2,"city"]
```

```python
df.iloc[0]
```

**Slicing**

```python
df.loc[1:3]
```

```python
df = pd.DataFrame({
    "name": ['Arun','Muhaz', 'Arjun']},
    index = ['A','B','C']
)

df.loc["B":"C"]
```

**.loc with Filtering**

```python
df = pd.DataFrame({
    "name": ['A','B','C','D'],
    "marks": [88,76,92,98],
    "city": ["Del","Pun","Koc","Mum"]
})

df.loc[df["marks"]>90]
```

**.iloc**

```python
df.iloc[0]
```

```python
"""
While performing slicing .iloc exclude the stop index
but .loc includes it.
"""
df.iloc[1:3] # slicing
```

**Row + Column**

```python
df.iloc[0,2] # df.iloc[row_index, col_index]
```

```python
df["name"]=="A"
```

```python
df[(df["name"] == 'A') & (df["city"] == 'Del')]
```

```python
df[(df["name"] == 'A') | (df["name"] == 'B')]
```

```python
df[~(df["name"] == 'B')]
```

**.isin()**

```python
df["name"].isin(['A','C'])
```

```python
df[df["name"].isin(['A','C'])]
```

**.between**

```python
df[df["marks"].between(90,100)]
```

**Sorting**<br>

**groupby() and aggregate functions**

```python
a = df[df["name"]>'B']["marks"].mean()
a
```

```python
df.groupby("name")["marks"].mean()
```

```python
df.groupby("name")["marks"].count()
```

```python
df.groupby(["name","city"]).count()
```

```python
df.groupby(["name","city"])["marks"].count()
```
